```python
import glob
import os
from access_nri_intake.source import builders
```

```python
from dask.distributed import Client
client = Client()
client
```

```python
# Directory containing the MOM6 experiment folders (test*-*)
BASE_PATH = "/g/data/nm03/ae7501/mom6_outputs/from_scratch/idealized"
```

```python
# Output folder for the datastores
OUT_DIR = os.path.join(os.getcwd(), "intake_style_datastore")   # or give a full path
os.makedirs(OUT_DIR, exist_ok=True)
```

```python
for test_num in range(263, 275):
    pattern = os.path.join(BASE_PATH, f"test{test_num}-*")
    matches = glob.glob(pattern)

    if not matches:
        print(f"WARNING: No folder found for test{test_num}, skipping...")
        continue

    if len(matches) > 1:
        print(f"WARNING: Multiple folders found for test{test_num}: {matches}, using first match.")

    path = matches[0]
    name = f"test{test_num}"
    description = f"Example datastore for test{test_num} ({os.path.basename(path)})"

    print(f"Processing {name} at {path}...")

    try:
        builder = builders.Mom6Builder(path=path).build()
        builder.save(name=name, description=description, directory=OUT_DIR)
        print(f"  Saved {name} -> {os.path.join(OUT_DIR, name + '.json')}")
    except Exception as e:
        print(f"  ERROR processing {name}: {e}")
```
